```python
!pip install polars
import polars as pl
from google.colab import drive
drive.mount('/content/drive')
events = pl.scan_parquet('/content/drive/MyDrive/all_events.parquet')
matches = pl.scan_parquet('/content/drive/MyDrive/matches.parquet')
leagues = pl.scan_parquet('/content/drive/MyDrive/leagues.parquet')
formations = pl.scan_parquet('/content/drive/MyDrive/formations.parquet')
print(events.collect_schema())
print(matches.collect_schema())
print(leagues.collect_schema())
preview_events = events.head(10).collect()
preview_leagues = leagues.head(10).collect()
preview_matches = matches.head(10).collect()
display(preview_events)
display(preview_leagues)
display(preview_matches)
```

```python
# Join matches and events using match IDs
combined_data = matches.join(events, left_on="wyId", right_on="matchId")

# Get one record per team per match
unique_team_matches = combined_data.unique(subset=["wyId", "team_id"])

# Identify if the team using the formation won the match
results = unique_team_matches.select([
    pl.col("team_formation").alias("formation"),
    (pl.col("winner") == pl.col("team_id")).alias("is_win")
])

# Group by formation and calculate the totals
formation_stats = results.group_by("formation").agg([
    pl.count("is_win").alias("total_games"),
    pl.col("is_win").sum().alias("total_wins")
])

# Calculate win rate
formation_stats = formation_stats.with_columns(
    (pl.col("total_wins") / pl.col("total_games")).alias("win_rate")
)

# Sort to see the most succesful formations at the top
sorted_stats = formation_stats.sort("win_rate", descending=True)
print(sorted_stats.collect())
```

```python
# Combine the datasets
formation_analysis = combined_data.group_by(["wyId", "team_id", "team_formation"]).agg([
    # Calculate performance accounting for mid-game switces by making windows(grouping by match, team, and formation.)
    pl.col("shot_xg").sum().alias("total_xG_created"),
    pl.col("shot_isGoal").sum().alias("actual_goals"),
    pl.count("label").alias("event_volume")
])

# Use across the whole season, see how each formation performs across all teams and matches.
global_formation_stats = formation_analysis.group_by("team_formation").agg([
    pl.count("wyId").alias("times_used"),
    pl.col("total_xG_created").mean().alias("avg_xG_per_window"),
    pl.col("actual_goals").sum().alias("total_goals_scored")
])

# 3. Filter out any formation that hasn't been used in at least 30 matches
# Calculate efficiency
significant_formations = global_formation_stats.filter(pl.col("times_used") > 30)

sorted_depth_stats = significant_formations.with_columns(
    (pl.col("total_goals_scored") / pl.col("times_used")).alias("goals_per_appearance")
).sort("avg_xG_per_window", descending=True)

print(sorted_depth_stats.collect())
```

```python
all_data = matches.join(events, left_on="wyId", right_on="matchId")
all_data_with_scores = all_data.with_columns(
    # High-Impact Positives
    pl.when(pl.col("shot_isGoal") == True).then(10.0)
    .when(pl.col("shot_onTarget") == True).then(3.0)

    # Mid-Impact Positives
    .when(pl.col("groundDuel_recoveredPossession") == True).then(0.5)
    .when((pl.col("type_primary") == "pass") &
          (pl.col("type_secondary").list.contains("progressive_pass")) &
          (pl.col("pass_accurate") == True)).then(0.4)
    .when(pl.col("groundDuel_progressedWithBall") == True).then(0.3)

    # Low-Impact Positives
    .when(pl.col("groundDuel_keptPossession") == True).then(0.1)
    .when(pl.col("pass_accurate") == True).then(0.05)

    # High-Impact Negatives
    .when(pl.col("infraction_redCard") == True).then(-15)
    .when(pl.col("infraction_yellowCard") == True).then(-1.5)

    # Low-Impact Negatives
    .when(pl.col("groundDuel_stoppedProgress") == True).then(-0.3)
    .when(pl.col("pass_accurate") == False).then(-0.2)

    # Neutral/Uncategorized actions get a 0
    .otherwise(0.0)
    .alias("action_score")
)

# 2. Calculate Performance Metrics per Window
formation_analysis = all_data_with_scores.group_by(["wyId", "team_id", "team_formation"]).agg([
    pl.col("shot_xg").sum().alias("total_xG_created"),
    pl.col("shot_isGoal").sum().alias("actual_goals"),
    pl.count("wyId").alias("event_volume"), # Changed from "label" since that wasn't in the new schema

    pl.col("action_score").sum().alias("total_match_rating"),
    pl.col("action_score").mean().alias("avg_action_rating_match")
])

# 3. Global Aggregation for the Season
global_stats = formation_analysis.group_by("team_formation").agg([
    pl.count("wyId").alias("times_used"),
    pl.col("total_xG_created").mean().alias("avg_xG_per_window"),
    pl.col("actual_goals").sum().alias("total_goals_scored"),

    pl.col("avg_action_rating_match").mean().alias("overall_play_rating"),
    pl.col("total_match_rating").mean().alias("avg_total_rating_per_match")
])

# 4. Efficiency and Final Reporting
final_report = global_stats.filter(pl.col("times_used") > 30).with_columns([
    (pl.col("total_goals_scored") / pl.col("times_used")).alias("goals_per_appearance")
]).sort("overall_play_rating", descending=True)

print(final_report.collect())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Collect the LazyFrame into a DataFrame and sort by 'times_used' descending
final_report_df = final_report.sort("times_used", descending=True).collect()

# Convert to Pandas format specifically for plotting compatibility with Seaborn/Matplotlib
df_plot = final_report_df.to_pandas()

cutoff_index = 10

# Filter the data down to our final formations list
filtered_df = df_plot.iloc[:cutoff_index]

# 3. Create the visualization
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Plot all formations to show the full trend and dropoffs
sns.barplot(
    x="times_used",
    y="team_formation",
    data=df_plot,
    palette="Blues_r"
)

# Draw a red vertical dashed line marking the calculated cutoff point
# We grab the 'times_used' value at the cutoff boundary to place our line
cutoff_val = df_plot['times_used'].iloc[cutoff_index - 1]
plt.axvline(x=cutoff_val, color="red", linestyle="--", linewidth=2, label=f"Cutoff Line (Top {cutoff_index})")

# Formatting elements
plt.title("Formation Usage Frequency & Selected Cutoff Threshold", fontsize=14, fontweight='bold')
plt.xlabel("Times Used", fontsize=12)
plt.ylabel("Team Formation", fontsize=12)
plt.legend(loc="lower right")
plt.tight_layout()

# Render the final graph in Colab
plt.show()

# 4. Display the filtered dataframe of formations you will look at
print(f"\n--- Formations Selected for Further Analysis (Top {cutoff_index}) ---")
print(filtered_df[["team_formation", "times_used", "overall_play_rating"]])
```
